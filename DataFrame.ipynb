## Part D – Cleaning a campus survey (missing values)
The Student Union asked 7 students from different Douala neighbourhoods how much they spend **per week** on transport and data bundles (FCFA). Some students skipped questions, so the data has **missing values (`NaN`)**. Run the cell below.

```python
survey = pd.DataFrame({
    "student":          ["Ange", "Blaise", "Carine", "Dieudonne", "Estelle", "Franck", "Grace"],
    "neighbourhood":    ["Bonamoussadi", "Akwa", None, "Bepanda", "Makepe", "Deido", "Logbessou"],
    "transport_fcfa":   [500, np.nan, 300, np.nan, 700, 400, 600],
    "data_bundle_fcfa": [2000, 1500, np.nan, 3000, np.nan, 2500, 1000]})
survey
```

### Question 23  *(3 marks)*
**TASK:** Where are the gaps? (a) A True/False table of missing cells (`missing_table`); (b) the **number of missing values per column** (`missing_count`); (c) the **percentage** missing per column (`missing_pct`).

> 💡 **HINT:** Slides 100–101: `df.isnull()` then `.sum()`. For a percentage divide by `len(df)` and multiply by 100.

```python
# ===== Q23 =====
# TASK: Where are the gaps? (a) A True/False table of missing cells (`missing_table`); (b) the **number of missing val
# missing_table -> isnull()
# missing_count -> isnull().sum()
# missing_pct   -> missing_count / len(survey) * 100
missing_table = survey.isnull()
missing_count = survey.isnull().sum()
missing_pct = survey.isnull().sum() / len(survey) * 100

# show your answers
print('missing_table'); display(missing_table)
print('missing_count'); display(missing_count)
print('missing_pct'); display(missing_pct)
```

### Question 24  *(3 marks)*
**TASK:** (a) Show the students that have **at least one** missing value (`incomplete`). (b) Show only the **complete** students (`complete`). (c) How many cells in the whole table are **not** missing (`n_valid`)?

> 💡 **HINT:** `df.isnull().any(axis=1)` gives one True/False per row; `~` flips True/False. `df.notnull().sum().sum()` totals the non-missing cells.

```python
# ===== Q24 =====
# TASK: (a) Show the students that have **at least one** missing value (`incomplete`). (b) Show only the **complete** 
# condition = rows having any missing value
# incomplete -> filter with the condition
# complete   -> filter with the opposite (~) condition
# n_valid    -> notnull() summed twice
incomplete = survey[survey.isnull().any(axis=1)]
complete = survey[~survey.isnull().any(axis=1)]
n_valid = survey.notnull().sum().sum()

# show your answers
print('incomplete'); display(incomplete)
print('complete'); display(complete)
print('n_valid'); display(n_valid)
```

### Question 25  *(2 marks)*
**TASK:** **Drop strategy.** (a) Remove every row that contains a missing value (`dropped_rows`). (b) Remove every *column* that contains a missing value (`dropped_cols`). Which one loses less information here?

> 💡 **HINT:** Slides 104–107: `dropna()` drops rows by default; `axis=1` drops columns.

```python
# ===== Q25 =====
# TASK: **Drop strategy.** (a) Remove every row that contains a missing value (`dropped_rows`). (b) Remove every *colu
# dropped_rows -> dropna()
# dropped_cols -> dropna(axis=1)
dropped_rows = survey.dropna()
dropped_cols = survey.dropna(axis=1)

# show your answers
print('dropped_rows'); display(dropped_rows)
print('dropped_cols'); display(dropped_cols)
```

**Answer to the comparison:** Dropping columns loses less information here because it removes only the columns containing missing values, while `dropna()` removes every row with at least one missing value.
